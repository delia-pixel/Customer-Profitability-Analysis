```python
df["calculated_net_sales"] = (
    df["gross_sales"] - df["discount_amount"] + df["tax_amount"]+ df["shipping_cost"]
)
df[[
    "gross_sales",
    "discount_amount",
    "tax_amount",
    "net_sales",
    "shipping_cost",
    "calculated_net_sales"
]].head(10)
```

```python
calculated_net_sales = (
    df["gross_sales"]
    - df["discount_amount"]
    + df["tax_amount"]
    + df["shipping_cost"]
)

difference = df["net_sales"] - calculated_net_sales

print("Mismatches:", (difference.abs() > 0.01).sum())
print("Maximum difference:", difference.abs().max())
```

```python
df["calculated_profit"] = df["net_sales"] - df["product_cost"]- df["shipping_cost"]
df[["net_sales",
    "product_cost",
    "profit",
    "calculated_profit"
]].head(10)
```

```python
calculated_margin = (
    df["profit"] / df["net_sales"]
) * 100

difference = (
    df["profit_margin_percentage"] - calculated_margin
)

print("Mismatches:", (difference.abs() > 0.01).sum())
print("Maximum difference:", difference.abs().max())
```

```python
actual_orders = df.groupby("customer_id")["order_id"].nunique()

recorded_orders = df.groupby("customer_id")["customer_order_count"].first()

comparison = pd.DataFrame({
    "actual_order_count": actual_orders,
    "recorded_order_count": recorded_orders
})

mismatches = comparison[
    comparison["recorded_order_count"] != comparison["actual_order_count"]
].copy()

display(mismatches.head())
```

```python
df_clean = df.copy()
```

```python
df_clean.shape
```

```python
df_clean.columns.tolist()
```

```python
df_clean.drop(
    columns=["calculated_net_sales", "calculated_profit"],
    errors="ignore",
    inplace=True
)
```

```python
df_clean.shape
```

```python
df_clean["return_status"] = df_clean["return_status"].fillna("No Return")
df_clean["return_status"].value_counts()
```

```python
df_clean["return_reason"] = df_clean["return_reason"].fillna("No Return")
df_clean["return_reason"].value_counts()
```

```python
df_clean["customer_rating"] = df_clean["customer_rating"].fillna("No Rating")
df_clean["customer_rating"].value_counts(dropna=False)
```

```python
df_clean["review_sentiment"] = df_clean["review_sentiment"].fillna("No Review")
df_clean["review_sentiment"].value_counts(dropna=False)
```

```python
df_clean["customer_review"] = df_clean["customer_review"].fillna("No Review")
df_clean["customer_review"].value_counts()
```

```python
df_clean["campaign_name"] = df_clean["campaign_name"].fillna("No Campaign Associated")
df_clean["campaign_name"].value_counts()
```

```python
df_clean["coupon_code"] = df_clean["coupon_code"].fillna("No Coupon Used")
df_clean["coupon_code"].value_counts()
```

```python
df_clean["delivery_status"].value_counts()
```

```python
pd.crosstab(
    df_clean["delivery_status"],
    df_clean["delivery_days"].isnull()
)
```

```python
df_clean["delivery_days"] = df_clean["delivery_days"].fillna("Not Delivered ")
df_clean["delivery_days"].value_counts()
```

```python
pd.crosstab(
    df_clean["delivery_status"],
    df_clean["estimated_delivery_days"].isnull()
)
```

```python
df_clean["estimated_delivery_days"] = (
    df_clean["estimated_delivery_days"].fillna("Not Applicable")
)
df_clean["estimated_delivery_days"].value_counts()
```

```python
df_clean["order_date"] = pd.to_datetime(df_clean["order_date"])
df_clean["order_date"].dtype
```

```python
df_clean["order_time"] = pd.to_datetime(
    df_clean["order_time"], format="%H:%M:%S"
).dt.time
df_clean["order_time"].dtype
```

```python
df_clean["order_date"].head()
```

```python
df_clean["customer_postal_code"] = df_clean["customer_postal_code"].astype("string")
df_clean["customer_postal_code"].dtype
```

```python
df_orderItems = pd.read_csv("../Datasets/order_items.csv")

df_orderItems.columns.tolist()
```

```python
calculated_gross_sales = (
    df_orderItems["quantity"] * df_orderItems["unit_price"]
)

mismatches = (
    (calculated_gross_sales - df_orderItems["gross_sales"]).abs() > 0.01
)

print("Total rows:", len(df_orderItems))
print("Incorrect gross sales:", mismatches.sum())
print("Correct gross sales:", (~mismatches).sum())
```
