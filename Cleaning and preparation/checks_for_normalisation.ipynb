```python
def check_separators(df):
    separators = [",", ";", "|"]

    for col in df.columns:
        for sep in separators:
            count = df[col].astype("string").str.contains(
                sep, regex=False, na=False
            ).sum()

            if count > 0:
                print(f"{col} | '{sep}' | {count} cells")

    print("Separator check completed. No seperator found")


```

checking all the other orders

```python
# Calculate running totals within each order
items_check = df.copy()

items_check["running_quantity"] = (
    items_check.groupby("order_id")["quantity"].cumsum()
)

items_check["running_gross_sales"] = (
    items_check.groupby("order_id")["gross_sales"]
    .cumsum()
    .round(2)
)

# Attach the original order totals
items_check = items_check.merge(
    ecommerce[["order_id", "quantity", "gross_sales"]],
    on="order_id",
    suffixes=("_item", "_order"),
    validate="many_to_one"
)

# Check whether a prefix of items matches both totals
items_check["prefix_matches"] = (
    (items_check["running_quantity"] == items_check["quantity_order"])
    &
    (
        (items_check["running_gross_sales"]
         - items_check["gross_sales_order"]).abs() <= 0.01
    )
)

matched_order_ids = set(
    items_check.loc[
        items_check["prefix_matches"], "order_id"
    ]
)

print(
    "Mismatched orders explained by an original prefix:",
    len(quantity_ids & matched_order_ids)
)

print(
    "Mismatched orders not explained:",
    len(quantity_ids - matched_order_ids)
)
```

```python
# Find the position where each order's original totals match
prefix_end = (
    items_check[items_check["prefix_matches"]]
    .groupby("order_id")
    .head(1)
    .index
)

# Identify additional records after the matching prefix
items_check["item_position"] = (
    items_check.groupby("order_id").cumcount()
)

prefix_positions = (
    items_check[items_check["prefix_matches"]]
    .groupby("order_id")["item_position"]
    .min()
)

items_check["prefix_end_position"] = (
    items_check["order_id"].map(prefix_positions)
)

additional_items = items_check[
    items_check["item_position"]
    > items_check["prefix_end_position"]
]

print("Additional item records:", len(additional_items))

print(
    "Orders containing additional records:",
    additional_items["order_id"].nunique()
)
```

ecommerce and product_repeated columns

```python
# Select items belonging to the matching original groups
original_items = items_check[
    items_check["item_position"]
    <= items_check["prefix_end_position"]
].copy()

# Calculate the total product cost for each order
original_cost = (
    original_items.groupby("order_id")["product_cost"]
    .sum()
    .round(2)
)

# Compare with ecommerce
cost_check = ecommerce[
    ["order_id", "product_cost"]
].copy()

cost_check["calculated_product_cost"] = (
    cost_check["order_id"].map(original_cost)
)

cost_check["difference"] = (
    cost_check["product_cost"]
    - cost_check["calculated_product_cost"]
).round(2)

print(
    "Orders with matching product cost:",
    (cost_check["difference"].abs() <= 0.01).sum()
)

print(
    "Orders with different product cost:",
    (cost_check["difference"].abs() > 0.01).sum()
)

print(
    "Orders without a calculated cost:",
    cost_check["calculated_product_cost"].isna().sum()
)
```

```python
# Helper function to check functional dependencies
def check_dependency(df, determinant, dependent):
    result = (
        df.groupby(determinant, dropna=False)[dependent]
        .nunique(dropna=False)
    )

    violations = (result > 1).sum()

    print(
        f"{determinant} -> {dependent}: "
        f"{violations} violations"
    )


print("1. PRODUCT DEPENDENCIES")

check_dependency(
    products_3nf,
    "product_subcategory",
    "product_category"
)

check_dependency(
    products_3nf,
    "brand",
    "supplier"
)

check_dependency(
    products_3nf,
    "supplier",
    "brand"
)


print("\n2. ORDER DEPENDENCIES")

check_dependency(
    orders_3nf,
    "warehouse",
    "shipping_method"
)

check_dependency(
    orders_3nf,
    "shipping_method",
    "warehouse"
)

check_dependency(
    orders_3nf,
    "campaign_name",
    "marketing_channel"
)


print("\n3. ORDER ITEM DEPENDENCIES")

check_dependency(
    order_items_3nf,
    "product_id",
    "unit_price"
)

check_dependency(
    order_items_3nf,
    "product_id",
    "product_cost"
)
```

verufying the keys of the created tables

```python

from pathlib import Path

folder = Path("../Datasets/3NF_Tables")

customers_3nf = pd.read_csv(folder / "Customers_3NF.csv")
customer_metrics_3nf = pd.read_csv(folder / "CustomerMetrics_3NF.csv")
products_3nf = pd.read_csv(folder / "Products_3NF.csv")
orders_3nf = pd.read_csv(folder / "Orders_3NF.csv")
order_items_3nf = pd.read_csv(folder / "OrderItems_3NF.csv")
order_reviews_3nf = pd.read_csv(folder / "OrderReviews_3NF.csv")

print("All six tables loaded successfully!")
```

```python
tables = {
    "Customers": (customers_3nf, "customer_id"),
    "CustomerMetrics": (customer_metrics_3nf, "customer_id"),
    "Products": (products_3nf, "product_id"),
    "Orders": (orders_3nf, "order_id"),
    "OrderItems": (order_items_3nf, "order_item_id"),
    "OrderReviews": (order_reviews_3nf, "order_id")
}

print("PRIMARY KEY VALIDATION\n")

for name, (table, pk) in tables.items():
    print(f"{name}:")
    print("Total records:", len(table))
    print("Duplicate PKs:", table[pk].duplicated().sum())
    print("Missing PKs:", table[pk].isna().sum())
    print()

```

verify foreign keys

```python
relationships = [
    ("Orders → Customers",
     orders_3nf, "customer_id",
     customers_3nf, "customer_id"),

    ("CustomerMetrics → Customers",
     customer_metrics_3nf, "customer_id",
     customers_3nf, "customer_id"),

    ("OrderItems → Orders",
     order_items_3nf, "order_id",
     orders_3nf, "order_id"),

    ("OrderItems → Products",
     order_items_3nf, "product_id",
     products_3nf, "product_id"),

    ("OrderReviews → Orders",
     order_reviews_3nf, "order_id",
     orders_3nf, "order_id")
]

print("FOREIGN KEY VALIDATION\n")

for name, child, fk, parent, pk in relationships:
    invalid = (
        child[fk].isna()
        | ~child[fk].isin(parent[pk])
    ).sum()

    print(f"{name}: {invalid} invalid references")

print(
    "\nOrderItems missing primary keys:",
    order_items_3nf["order_item_id"].isna().sum()
)
```
